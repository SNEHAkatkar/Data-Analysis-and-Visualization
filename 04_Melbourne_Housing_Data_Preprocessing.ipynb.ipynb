```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv(r"C:\Users\Sneha katkar\OneDrive\Documents\melb_data.csv")
print("Original Dataset")
print(df.head())
print("\nshape of dataset:")
print(df.shape)
```

```python
print("Column Name:")
print(df.columns)
```

```python
print("Data info")
print(df.info)
```

```python
print("\nStatistical Summary:")
print(df.describe())
```

```python
df_mean=df.copy()
mean_building_area=df_mean["BuildingArea"].mean()
df_mean["BuildingArea"]=df_mean["BuildingArea"].fillna(mean_building_area)
print("\n mean BuildingArea :",mean_building_area)
print("\nMean BuildingArea values after mean Imputation:")
print(df_mean["BuildingArea"].isnull().sum())
```

```python
df_median = df.copy()

# Calculate the median value of YearBuilt
median_year_built = df_median["YearBuilt"].median()

# Fill missing values with the median
df_median["YearBuilt"] = df_median["YearBuilt"].fillna(median_year_built)

print("\nMedian Year Built:", median_year_built)
print("Missing YearBuilt values after Median Imputations:")
print(df_median["YearBuilt"].isnull().sum())
```

```python
df_ffill = df.copy()

# Use capitalized "Car" on both sides
df_ffill["Car"] = df_ffill["Car"].ffill()

print("Missing car values after Forward Fill :")
print(df_ffill["Car"].isnull().sum())
```

```python
comparision=pd.DataFrame({
    "Original BuildingArea": df["BuildingArea"].head(15),
    "Mean Imputed BuildingArea":df_mean["BuildingArea"].head(15),
    
    "Original YearBuilt": df["YearBuilt"].head(15),
    "Mean Imputed YearBuilt":df_median["YearBuilt"].head(15),
    
    "Original Car": df["Car"].head(15),
    "Mean Imputed Car":df_ffill["Car"].head(15),
     
})
print("\nomparision of Imputation Methods:")
print(comparision)
```

```python
df_processed=df.copy()
council_mode=df_processed["CouncilArea"].mode()[0]
df_processed["CouncilArea"]=df_processed["CouncilArea"].fillna(council_mode)
print("\nCouncilArea Mode:",council_mode)
print("Missing Categorial Values After imputation ")
print(df_processed[["CouncilArea"]].isnull().sum())
```

```python
df_encoded=pd.get_dummies(
    df_processed,
    columns=["Type","Method"],
    dtype=int
)
print("\nDataset After ONe=Hot Encoding")
print(df_encoded.head())
```

```python
df_label=df.copy()
label_encoder=LabelEncoder()
df_label["CouncilArea_Encoded"]=label_encoder.fit_transform(
    df_label["CouncilArea"]
)
print("\nLabel Encoding of councilArea")
print(
    df_label[
        ["CouncilArea","CouncilArea_Encoded"]].head(10))
```
