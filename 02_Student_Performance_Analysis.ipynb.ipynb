```python
#data wrangling
grade_info=pd.DataFrame({'grade':['A','B','C','D','F'],
                         'grade description':['Excellent','Good','Average','Below Average','Fail']
                        })
print("\nGrade Information DataFrame:")
print(grade_info)
```

```python
#merging dataframes
merged=pd.merge(
df,
grade_info,
on='grade',
how='left'
)
print("\nMerged DataFrame:")
print(merged.head())
```

```python
#joining DataFrame
grade_details=pd.DataFrame({
    'description':[
        'Excellent','Good','Average','Below Average','Fail'],
    'performance_level':[
        'Very High','High','Medium','low','Very Low'
         ]
},index=['A','B','C','D','F'])
print(grade_details.head())

grade_counts=df['grade'].value_counts().to_frame('student_count')
print(grade_counts.head())
```

```python
joined=grade_counts.join(grade_details)
print(joined)
```
